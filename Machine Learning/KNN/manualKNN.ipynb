```python
import numpy as np
```

```python
def distanceFromOnePoint(point1, point2):
    dx = point1[0] - point2[0]
    dy = point1[1] - point2[1]
    dz = point1[2] - point2[2]

    return np.sqrt(dx**2 + dy**2 + dz**2)
```

```python
def foundLabel(kPoints):
    labels = {}
    for point in kPoints:
        label = point[3]
        if label not in labels:
            labels[label] = 0
        labels[label] += 1

    return max(labels, key=labels.get)
```

```python
def knn(data, pointToFound, k):
    distanceFromData = []
    for dataPoint in data:
        dist = distanceFromOnePoint(dataPoint, pointToFound)
        distanceFromData.append((dist, dataPoint))
    distanceFromData.sort(key=lambda x: x[0])

    kNearestPoints = [item[1] for item in distanceFromData[:k]]
    print(foundLabel(kNearestPoints))
```
