```python
#open you csv file and print the first 5 rows of the file
arr = np.genfromtxt("grades.csv", delimiter=',')
```

<p dir = 'rtl'>
הקובץ כפי שהזכרנו קודם, מייצג הישגי תלמידים במספר מבחנים. כל שורה מייצג תלמיד</br>
מצאו כמה תלמידים יש בכיתה ובכמה מקצועות התלמידים נבחנו
</p>

```python
# print how many students there are 
# print how many exams there were
print(arr.shape[0])
print(arr.shape[1])
```

<p dir = 'rtl'>
מצא מה הציון הכי גבוה שכל תלמיד קיבל. </br>
מצא מה הציון הכי נמוך שכל תלמיד קיבל</br>
מצא עבור כל מקצוע מה היה הציון הכי גבוה והכי נמוך
</p>

```python
#find the greatest grades each student got 
#find the lowest grade each student got
#find the highest grade in each exam
#find the lowest exam
print(np.max(arr,axis=1))
print(np.min(arr,axis=1))

print(np.max(arr,axis=0))
print(np.min(arr,axis=0))
```

<p dir = 'rtl'>
כעת מצא את הממוצע הכללי של כל המבחנים</br>
מצא את ממוצע הציונים של כל תלמיד 
ומה הממוצע של כל המבחנים
</p>


```python
#find avg
print(np.average(arr)) 
print(np.average(arr, axis=0)) # exams
print(np.average(arr, axis=1)) # per student
```

<p dir = 'rtl'>
נבחר מקצוע מספר 3 ונתתחיל לבחון אותה </BR>
תחילה נמיין את הציונים במקצוע ונדפיס אותם.</BR>
לאחר מכן, נמצא את החציון של הבחינה. 
</p>

```python
third_exam = arr[:, 3]
third_exam_sorted = np.sort(third_exam)

print(third_exam_sorted)

print(np.median(third_exam_sorted))
```

<p dir = 'rtl'>
 קלוט ציון ומצא כמה תלמידים קיבלו מעל הציון הזה בכל מקצוע. </br>
 (רמז: יש הוראה ב- numpy  שנקרא : np.count_nonzero())
 </p>

```python
g = int(input())

np.count_nonzero(arr > g, axis=0)
```

<p dir = 'rtl'>
לסיכום: מצא את המבחן שבה התלמיד קיבל את הציון הצטיין בה, המקצוע בה התלמידים קיבלו את ציון הכי גבוה</br>
</p>


```python
# find the indices of the subject with the best acheivements and the one most "even"
indices = np.argmax(arr, axis=1)
indices = indices.reshape(-1, 1)
print(indices)
```
