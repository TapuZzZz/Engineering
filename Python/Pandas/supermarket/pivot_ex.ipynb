```python
import pandas as pd
import numpy as np
```

```python
#יש לטעון את רשימת הקיניות מהקובץ - האינדקס הוא מספר ההזמנה
df = pd.read_csv('supermarket_sales.csv', index_col=0)
```

```python
#הראו 5 שורות ראשונות
df.head(5)
```

```python
#הציגו את רשימת העמודות
df.columns
```

```python
#כתבו כמה ערים יש וכמה יש בכל עיר
df['City'].nunique()

df['City'].value_counts()
```

```python
#יש ליצור את טבלאת הפיבוט הבאה, שימו לב כל הערכים זה ממוצעים
pivot = df.pivot_table(
    values=['Quantity','Rating','Tax 5%','Total','Unit price','cogs','gross income','gross margin percentage'],
    index='City',
    aggfunc='mean'
)
pivot
```

```python
# ייצרו את טבלאת הפיבוט הבאה, שימו לב החישוב הוא סכום העמודות
pivot2 = df.pivot_table(
    values=['Rating','gross income'],
    index='City',
    aggfunc='sum'
)
pivot2
```

```python
#בצעו את אותה טבלא, אבל לעמודה הראשונה יש לחשב ערך מקסימלי ולשניה סטיית תקן
pivot3 = df.pivot_table(
    values=['Rating','gross income'],
    index='City',
    aggfunc={'Rating':np.std,'gross income':np.max}
)
pivot3
```

```python
#יש להראות בכל עיר את ממוצע המדד לגברים ןלנשים ואת ממוצע ההכנסה הכללית לגבר ונשים
#מדד- rating
#ממוצע הכנסה - gross income
pivot4 = df.pivot_table(
    values=['Rating','gross income'],
    index='City',
    columns="Gender",
    aggfunc='mean'
)
pivot4
```

```python
#יש להראות בכל מגדר את סטיית התקן של הכמות ומחיר יחידה בחתך כל עיר
pivot5 = df.pivot_table(
    values=['Quantity','Unit price'],
    index='City',
    columns="Gender",
    aggfunc=np.std
)
pivot5
```
