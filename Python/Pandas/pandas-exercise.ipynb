```python
# 6)
df["bill_ratio"] = df["bill_length_mm"] / df["bill_depth_mm"]
```

```python
# 7)
pivot = pd.pivot_table(
    df,
    values="body_mass_g",
    index="species",
    columns="sex",
    aggfunc="mean"
)
print(pivot)
```

```python
# 8)
pivot2 = pd.pivot_table(
    df,
    values="body_mass_g",
    index=["species", "island"],
    columns="sex",
    aggfunc="mean",
    margins=True
).round(1)
print(pivot2)
```

```python
# 9)
pivot3 = pd.pivot_table(
    df,
    values="flipper_length_mm",
    index="species",
    columns="sex",
    aggfunc=["mean", "count"],
).round(1)
print(pivot3)
```

```python
# 10)
df["size_category"] = pd.cut(
    df["body_mass_g"],
    bins=3,
    labels=["Light", "Medium", "Heavy"]
)

pivot4 = pd.pivot_table(
    df,
    values="species",
    index=["island", "size_category"],
    columns="sex",
    aggfunc="count",
    fill_value=0
)
print(pivot4)
```
