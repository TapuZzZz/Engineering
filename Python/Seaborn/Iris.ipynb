<div dir="rtl">
נבדוק את אורך עלה הכותרת אצל כל סוג בעזרת גרף קופסה
</div>

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset('iris')

sns.boxplot(data=df, x='species', y='petal_length', hue='species', legend=False)
plt.ylabel('Petal length')
plt.xlabel('')
plt.title('אורכי ה patels לפי הסוגים השונים')
plt.show()
```

<div dir="rtl">
המסקנה שאני מסיק זה שאפשר לזהות את הסוג הכחול יחסית בבירור, אך לעומת זאת בין הסוג הירוק לכתום יש חפיפה בטווחים דבר שטיפה יקשה במקרים מאוד מסויימים לדעת בוודאות מלאה
</div>

```python
sns.scatterplot(data=df, x='petal_width', y='petal_length', hue='species')
plt.ylabel('Petal length')
plt.xlabel('Petal width')
plt.title('יחס בין אורך לרוחב ה petals')
plt.show()
```

<div dir="rtl">
ניתן לראות כי היחס בין אורך ה petal לרוחב שלו נותן לנו הבנה שהקבוצות מבודדות וכי ברוב המקרים ניתן להסיק כי מתחת ל 1 הסוג הכחול, בין 1 ל 1.7 הסוג הכתום ומעל ירוק
</div>

```python
sns.kdeplot(data=df, x='petal_width', hue='species')
plt.xlabel('Petal width')
plt.title('התפלגות אורכי ה petals לפי הסוגים השונים')
plt.show()
```

<div dir="rtl">
גם פה אפשר לראות את ההבדלים בהם מתקיימים הספייקים שמציגים את אורך הpetal
</div>

```python
sns.pairplot(data=df, hue='species', diag_kind='hist')
```
